# LAB | Feature Engineering
**Solved version — Spaceship Titanic**

Same dataset as the intro lab, but this time we prepare the categorical columns properly
(encoding, cleaning `Cabin`, dropping useless columns) before training KNN.

**Load the data**

https://raw.githubusercontent.com/data-bootcamp-v4/data/main/spaceship_titanic.csv

```python
# Libraries
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
```

```python
spaceship = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/spaceship_titanic.csv")
spaceship.head()
```

**Check the shape of your data**

```python
spaceship.shape
```

**Check for data types**

```python
spaceship.dtypes
```

**Check for missing values**

```python
spaceship.isna().sum()
```

There are multiple strategies to handle missing data

- Removing all rows or all columns containing missing data.
- Filling all missing values with a value (mean in continouos or mode in categorical for example).
- Filling all missing values with an algorithm.

For this exercise, because we have such low amount of null values, we will drop rows containing any missing value.

```python
spaceship = spaceship.dropna()
spaceship.shape
```

- **Cabin** is too granular - transform it in order to obtain {'A', 'B', 'C', 'D', 'E', 'F', 'G', 'T'}

`Cabin` looks like `B/0/P` (deck / number / side). The deck (the first letter) is the useful part.
`.str[0]` keeps just that first character, turning hundreds of unique cabins into 8 deck categories.

```python
spaceship["Cabin"] = spaceship["Cabin"].str[0]
print(sorted(spaceship["Cabin"].unique()))
spaceship["Cabin"].value_counts()
```

- Drop PassengerId and Name

`PassengerId` and `Name` are unique per passenger — they carry no pattern the model can learn from,
so we drop them.

```python
spaceship = spaceship.drop(columns=["PassengerId", "Name"])
spaceship.head()
```

- For non-numerical columns, do dummies.

**One-hot encoding.** `pd.get_dummies` turns each categorical column into 0/1 columns, one per
category. KNN needs numbers, so this is required. `drop_first=True` removes one category per column to
avoid redundancy (if it's not any of the others, it must be the dropped one).

```python
spaceship = pd.get_dummies(spaceship, drop_first=True)
print(spaceship.shape)
spaceship.head()
```

**Perform Train Test Split**

```python
X = spaceship.drop(columns=["Transported"])
y = spaceship["Transported"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=17
)

X_train.shape, X_test.shape
```

**Model Selection**

In this exercise we will be using **KNN** as our predictive model.

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier()   # default n_neighbors=5
knn.fit(X_train, y_train)
```

- Evaluate your model's performance. Comment it

```python
accuracy = knn.score(X_test, y_test)
print("accuracy:", round(accuracy, 4))
```

```python
from sklearn.metrics import confusion_matrix, classification_report

pred = knn.predict(X_test)
print(confusion_matrix(y_test, pred))
print()
print(classification_report(y_test, pred))
```

**Comment.**

The model reaches about **80% accuracy**. Compared to the intro lab (where we used only the numeric
columns and got ~78%), adding the encoded categorical columns — `HomePlanet`, `CryoSleep`, `Cabin`
deck, `Destination`, `VIP` — improved the result slightly.

The dataset is fairly balanced (~50% Transported), so accuracy is a reasonable metric here, and
precision and recall are similar for both classes.

**How to push it further:** scale the numeric columns (KNN is distance-based, and `Spa` / `FoodCourt`
reach thousands while `Age` stays under 80), and tune `n_neighbors` instead of leaving it at the
default 5.
